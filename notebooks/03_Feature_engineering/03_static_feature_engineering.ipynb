# 03. Static Feature Engineering
This notebook documents the static spatial feature extraction pipeline (GEE NDVI buffers, GEE ESA land cover fractions, and OSMnx road distance/density).

```python
from pathlib import Path
import pandas as pd

ROOT = Path("../../")
PROC_DIR = ROOT / "data/processed"
PROC_DIR.mkdir(parents=True, exist_ok=True)
```

## 1. Description of GEE & OSMnx Pipeline
Because Google Earth Engine (GEE) and OSMnx calculations require active authentication and are slow, we keep a documented copy of the code and read the pre-computed outputs from `station_spatial_features.csv` to ensure consistent and reproducible datasets.

### Land Cover Fractions (ESA WorldCover)
We extract fraction values of tree, shrubland, grassland, cropland, built-up, bare, and water in 500m, 1km, 2km, and 5km buffers.

### OSM Road Features
We extract minimum distance to motorway, primary, secondary, and any major road, along with road density (1km/2km) and number of major road intersections (5km).

```python
spatial_features_path = PROC_DIR / "station_spatial_features.csv"
if spatial_features_path.exists():
    df_sf = pd.read_csv(spatial_features_path)
    df_sf.to_csv(PROC_DIR / "03_station_spatial_features.csv", index=False)
    print(f"Saved static spatial features. Shape: {df_sf.shape}")
else:
    print("Error: station_spatial_features.csv not found!")
```
